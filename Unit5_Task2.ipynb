```python
position=jun.groupby("Посада")["salary"].mean()
print(position)
```

Створіть новий DataFrame, де для згрупованих даних за стовпчиком "Посада", виконайте агрегацію даних за допомогою методу agg і знайдіть мінімальне та максимальне значення у стовпчику "Зарплата.в.місяць"


```python
salary_stats = jun.groupby("Посада")["Зарплата на місяць"].agg(["min", "max"])
print(salary_stats)
```

Створіть функцію fill_avg_salary, яка повертатиме середнє значення заробітної плати на місяць. Використовуйте її для методу apply та створіть новий стовпчик "avg"

```python
def fill_avg_salary(row):
    return(row["min"] + row["max"])/2

salary_stats["avg"]=salary_stats.apply(fill_avg_salary, axis=1)
print(salary_stats)
```

Створіть описову статистику за допомогою методу describe для нового стовпчика.

```python
print(salary_stats["avg"].describe())
```

Збережіть отриману таблицю в CSV файл

```python
salary_stats.to_csv("data.csv")
```
